Parsing students information with pydantic


```python
from pydantic import BaseModel, Field

class Learner(BaseModel):
  registration_number: str
  full_name: str
  faculty: str
  department: str
  year_of_study: int
  student_email: str
```

```python
Student = Learner(
    registration_number= "ENE212-0088/2023",
    full_name= "Branice Nafula",
    faculty= "College of Engineering and Technology",
    department= "Electronics and Electrical Engineering",
    year_of_study= 3,
    student_email= "nafula.branice@students.jkuat.ac.ke"
)
```

```python
print(Student.registration_number)
print(Student.full_name)
```

Workflow queries system

```python
class StudentQueryResponse(BaseModel):
  intent: str
  urgency: str= Field(pattern= "^(low|medium|high)$")
  suggested_coordinator: str
  copy_paste_reply: str
```

```python
Response_output ='{"intent": "attachment inquiry", "urgency": "medium", "suggested_coordinator": "MR Mureu", "copy_paste_reply": "Hello, submit your signed logbook before Friday"}'
```

```python
extracted_data = StudentQueryResponse.model_validate_json(Response_output)
print("--- Structured Workflow Dashboard ---")
print(f"Assign Ticket To: {extracted_data.suggested_coordinator}")
print(f"Intent: {extracted_data.intent}")
print(f"Urgency: {extracted_data.urgency}")
print(f"Reply: {extracted_data.copy_paste_reply}")
```
